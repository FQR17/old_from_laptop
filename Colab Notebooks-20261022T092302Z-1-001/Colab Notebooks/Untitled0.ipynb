```python
import random

def сaesar(massage, step, chars, length, cipher):
    outMassage = ''
    counterMassage = 0
    if not cipher:
        step = -step
    for i in massage:
        counterChars = 0
        for j in chars:
            if chars[counterChars] == massage[counterMassage]:
                if counterChars + step >= length:
                    counterChars -= length
                if counterChars + step <= -length:
                    counterChars += length
                outMassage += chars[counterChars + step]
            counterChars += 1
        if i not in chars:
          outMassage += i
        counterMassage += 1
    print(outMassage)

def randomChars(chars):
    array = []
    j = 0
    RandomChars = ''
    for i in chars:
        array.append(chars[j])
        j += 1
    length = len(array)
    for i in range(0, length):
        rnd = random.randint(0, length - 1)
        temp = array[i]
        array[i] = array[rnd]
        array[rnd] = temp
    j = 0
    for i in range(0,length):
        RandomChars += array[j]
        j += 1
    return RandomChars

def replacement(massage, chars, length, RandomChars, cipher):
    outMassage = ''
    counterMassage = 0
    if not cipher:
        box = chars
        chars = RandomChars
        RandomChars = box
    for i in range(0,len(massage)):
        j = 0
        while j < length:
            if massage[counterMassage] == chars[j]:
                outMassage += RandomChars[j]
            j += 1
        if massage[i] not in chars:
          outMassage += massage[i]
        counterMassage += 1
    print(outMassage)

def vigenere(massage, chars, length, key, cipher):
    outMassage = ''
    counterMassage = 0
    counterKey = 0
    addKey = key
    while True:
        if len(key) < len(massage):
            key += addKey
        else: break
    for i in range(0, len(massage)):
        counterInLine = 0
        for c in range(0,length):
            if chars[counterInLine] == massage[counterMassage]:
                counterInColumn = 0
                for j in range(0, length):
                    if chars[counterInColumn] == key[counterKey]:
                        if cipher:
                            if counterInLine + counterInColumn >= length:
                                counterInColumn -= length
                        if not cipher:
                            if counterInLine - counterInColumn <= 0:
                                outMassage += chars[counterInLine - counterInColumn + length]
                                break
                            else:
                                outMassage += chars[counterInLine - counterInColumn]
                                break
                        outMassage += chars[counterInLine + counterInColumn]
                    counterInColumn += 1
            counterInLine += 1
        if massage[i] not in chars:
          outMassage += massage[i]
          counterKey -= 1
        counterKey += 1
        counterMassage += 1
    print(outMassage)

def main():
    chars = 'abcdefghijklmnopqrstuvwxyz'
    length = len(chars)
    #print(length)
    step = 15
    RandomChars = randomChars(chars)
    while True:
        while True:
            comand = input('Что выхотите сделать? (encrypt/decipher) > ')
            if comand == 'encrypt':
                cipher = True
            elif comand == 'decipher':
                cipher = False
            else:
                print('Такой функции нет')
                break
            Massage = input('Введите сообщение шифровки > ')
            if Massage == '':
                break
            comand = input('Какой алгоритм используем? (caesar/replacement/vigenere) > ')
            if comand == 'caesar':
                сaesar(Massage, step, chars, length, cipher)
            elif comand == 'replacement':
                replacement(Massage, chars, length, RandomChars, cipher)
            elif comand == 'vigenere':
                key = input('Введите ключ шифровки > ')
                vigenere(Massage, chars, length, key, cipher)
            else:
                print('Такого алгоритма нет')
                break
main()
```
