```python
import random
import math
P = 10**(-7)
V = 3 #pass/min
T = 60*24*30 # 30 days
CountPassIn30Days = V*T # = 129600
S1 = CountPassIn30Days/P #S1 = 1296 * 10^9
A = 36
L = math.ceil(math.log(S1,A))
print(L)
'S = A^L = 2821109907456 => S1'
chars = 'abcdefghijklmnopqrstuvwxyz1234567890' # алфавит
length = L
password = ''
for i in range(length):
        password += random.choice(chars)
print(password)
```
