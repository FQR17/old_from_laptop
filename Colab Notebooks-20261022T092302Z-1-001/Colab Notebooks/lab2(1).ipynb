```python
import random

def main():
    matrix = [['Ivan', 0, 0, 0, 0, 0, 0],
              ['Sergey', 0, 0, 0, 0, 0, 0],
              ['Boris', 0, 0, 0, 0, 0, 0],
              ['Boris2', 0, 0, 0, 0, 0, 0],
              ['Boris3', 0, 0, 0, 0, 0, 0],
              ['Boris4', 0, 0, 0, 0, 0, 0],
              ['Boris5', 0, 0, 0, 0, 0, 0],
              ['Boris6', 0, 0, 0, 0, 0, 0],
              ['Boris7', 0, 0, 0, 0, 0, 0]]

    LenStringOfMatrix = 0
    for i in matrix[0]:
        LenStringOfMatrix += 1

    for i in range(len(matrix)):
        j = 1
        while j < LenStringOfMatrix:
            if matrix[i][0] == 'Ivan':
                matrix[i][j] = 3
            else:
                matrix[i][j] = random.randint(0, 3)
            j += 1
    while True:
        User = input('User: ')
        if User == 'exit':
          break
        for i in range(len(matrix)):
            if User == matrix[i][0]:
                print('Идентификация прошла успешно, добро пожаловать в систему')
                print('Перечень Ваших прав:')
                j = 1
                while j < LenStringOfMatrix:
                    if matrix[i][j] == 0:
                        print('Объект%d     Запрет' % (j))
                    elif matrix[i][j] == 1:
                        print('Объект%d     Чтение' % (j))
                    elif matrix[i][j] == 2:
                        print('Объект%d     Чтение, запись' % (j))
                    elif matrix[i][j] == 3:
                        print('Объект%d     Полные права' % (j))
                    else:
                        print('Объект%d   Ошибка доступк, обратитесь к администратору' % (j))
                    j += 1
                while True:
                    command = input('Жду ваших указаний (right, write, grant, quit) > ')
                    if command == 'grant':
                        while True:
                            object = input('Над каким объектом производится операция? ')
                            try:
                                object = int(object)
                                if 0 < object > LenStringOfMatrix-1:
                                    pass
                                else:
                                    break
                            except:
                                print('Введите корректный номер объекта')
                        if matrix[i][object] == 3:
                            Right = input('Какое право передается? (right, write, grant) ')
                            if Right == 'right':
                                Right = 1
                            elif Right == 'write':
                                Right = 2
                            elif Right == 'grant':
                                Right = 3
                            user = input('Какому пользователю передается право? ')
                            UserDetect = False
                            for j in range(len(matrix)):
                                if user == matrix[j][0]:
                                    UserDetect = True
                                    if matrix[j][object] < Right:
                                      matrix[j][object] = Right
                                    else:                                 
                                      print('Нельзя понизить права')
                            if UserDetect:
                                print('Операция прошла успешно')
                            else:
                                print('Пользователь не найден')
                        else:
                            print('Отказ в выполнении операции. У Вас нет прав для ее осуществления')

                    elif command == 'right':
                        while True:
                            object = input('Над каким объектом производится операция? ')
                            try:
                                object = int(object)
                                if 0 < object > LenStringOfMatrix-1:
                                    pass
                                else:
                                    break
                            except:
                                print('Введите корректный номер объекта')
                        if matrix[i][object] >= 1:
                            print('Операция прошла успешно')
                        else: print('Отказ в выполнении операции. У Вас нет прав для ее осуществления')

                    elif command == 'write':
                        while True:
                            object = input('Над каким объектом производится операция? ')
                            try:
                                object = int(object)
                                if 0 < object > LenStringOfMatrix-1:
                                    pass
                                else:
                                    break
                            except:
                                print('Введите корректный номер объекта')
                        if matrix[i][object] >= 2:
                            print('Операция прошла успешно')
                        else:
                            print('Отказ в выполнении операции. У Вас нет прав для ее осуществления')

                    elif command == 'quit':
                        print('Работа пользователя %s завершена. До свидания.'%(User))
                        break
main()
```
